### 1.upper():
Converts all letters to uppercase

```python
name="python"
name1="pthon is the best language"
print(name.upper())
print(name1.upper())
```

### 2.lower():
Converts all letters to lowercase

```python
name="PYTHON"
name1="PYTHON IS THE BEST LANGUAGE"
print(name.lower())
print(name1.lower())
```

### 3.capitalize():
Makes the first letter uppercase

```python
s="hello world"
s2="i like python"
print(s.capitalize())
print(s2.capitalize())
```

### 4.title():
Makes the first letter of every word uppercase

```python
s="hello world"
s2="i like python"
print(s.title())
print(s2.title())
```

### 5.strip():
Removes spaces from beginning and end

```python
s="     hello world     "
s2="    i like python   "
print(s.strip())
print(s2.strip())
```

### 6.replace():
Replaces one text with another

```python
s="hello world"
s2="i like java"
print(s.replace("world","friends"))
print(s2.replace("java","python"))
```

### 7.find():
Finds the position of a character/text

```python
s="hello world"
s2="i like python"
print(s.find("w"))
print(s2.find("t"))
```

### 8.count():
Counts occurrences of a character/text

```python
s="hello world"
s2="i like java"
print(s.count("l"))
print(s2.count("a"))
print(s.count("lo"))
print(s2.count("ja"))
```

### 9.startswith():
Checks whether a string starts with specific text

```python
s="hello friends"
s2="i like python"
print(s.startswith("hello"))
print(s2.startswith("il"))
print(s.startswith("he"))
print(s2.startswith("i l"))
```

### 10.endswith():
Checks whether a string ends with specific text

```python
s="hello friends"
s2="i like python"
print(s.endswith("friends"))
print(s2.endswith("on"))
print(s.endswith("end"))
print(s2.endswith("n"))
```

### 11.isalpha():
Checks whether all characters are alphabets

```python
s="hello world"
s2="i like python3.14.6"
print(s.isalpha())
print(s2.isalpha())
s="helloworld"
s2="ilikepython3.14.6"
print(s.isalpha())
print(s2.isalpha())
```

### 12. isdigit():
Checks whether all characters are digits

```python
s="12358"
s1="ilikepython3.14.6"
s2="123456789"
s3="3.14.6"
print(s.isdigit())
print(s1.isdigit())
print(s2.isdigit())
print(s3.isdigit())
```

### 13.isalnum():
Checks whether characters are alphabets or numbers

```python
s="iamshaikh234"
s1="ilikepython3.14.6"
s2="123456789"
s3="i like python"
s4="ilikepython"
print(s.isalnum())
print(s1.isalnum())
print(s2.isalnum())
print(s3.isalnum())
print(s4.isalnum())
```

### 14.split():
Splits a string into a list

```python
s="hello friends"
s1="i like python"
s2="hello friends 1122"
s3="i like python 3.14.6"
print(s.split())
print(s1.split())
print(s2.split())
print(s3.split())
```

### 15.join():
Joins elements into a string

```python
s=['hello', 'friends']
s1=['i', 'like', 'python']
s2=['hello', 'friends', '1122']
s3=['i', 'like', 'python', '3.14.6']
print("".join(s))
print(''.join(s1))
print(''.join(s2))
print(''.join(s3))
```
